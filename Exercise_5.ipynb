## Task 12 c

Concentration profile:
$$ c(z,t) = \frac{Q}{A \sqrt{2 \pi d D t}} \exp \left( -\frac{(z - z_0)^2}{2 d D t} \right) $$

Concentration standard deviation:
$$ \sigma(t) = \sqrt{2 D d t} $$

Half Width at Half Maximum (FWHM):
$$ \text{FWHM} = 2 \sqrt{2 \ln(2)} \, \sigma(t) $$

```python
D = 2.5e-9  # m^2/s
z0 = 0 # m
z_1 = 2.5e-2 # m
d = 1 # dimensionality

t_list = [100, 65*60, 2*60*60 + 47*60]  # s

def sigma(D,d,t):
    return np.sqrt(2*d*D*t)

for t in t_list:
    sigma_1 = sigma(D, d, t)
    print(f"At time t = {t} s, sigma = {sigma_1:.2e} m")
```

```python
def FWHM(sigma):
    return 2 * np.sqrt(2 * np.log(2)) * sigma

for t in t_list:
    sigma_1 = sigma(D, d, t)
    fwhm_1 = FWHM(sigma_1)
    print(f"At time t = {t} s, FWHM = {fwhm_1:.2e} m")
```

## Task 12 d

Concentration profile:
$$ c(z,t) = \frac{Q}{A \sqrt{2 \pi d D t}} \exp \left( -\frac{(z - z_0)^2}{2 d D t} \right) $$

Concentration Half-Time at position $z=0$:
$$ c(0,t_{1/2}) = 0.5 \cdot c(0,t) $$
$$ c(0,t_{1/2})/ c(0,t) = 0.5 = (t_{1/2}/t_0)^{-1/2}$$
$$ t_{1/2} = 4 t_0 $$

```python
D = 2.5e-9  # m^2/s
z0 = 0 # m
z_1 = 0 # m
d = 1 # dimensionality
A = 1e-8 # cross-sectional area in m^2
Q = 1e-9 # quantity of substance released in moles

def concentration_profile(A, Q, z0, z1, D, d, t):
    return (Q / (A * np.sqrt(2 * d * np.pi * D * t)) * np.exp(-((z1 - z0)**2) / (2 * d *  D * t)))

def concentration_half_time(A, Q, z0, z1, D, d, t):
    c_t = concentration_profile(A, Q, z0, z1, D, d, t)
    t_half = 4 * t
    c_t_half = concentration_profile(A, Q, z0, z1, D, d, t_half)
    return c_t_half

print("Concentration Profile Results:")
for t in t_list:
    conc_1 = concentration_profile(A, Q, z0, z_1, D, d, t)
    print(f"At time t = {t} s, concentration at z = {z_1} m is {conc_1:.2e} mol/m^3")
    
print("\nConcentration Half-Time Results:")
    
for t in t_list:
    conc_half_1 = concentration_half_time(A, Q, z0, z_1, D, d, t)
    conc_half_time = 4 * t
    print(f"At time t = {t} s, concentration half-time ({conc_half_time} s) at z = {z_1} m is {conc_half_1:.2e} mol/m^3")
```
